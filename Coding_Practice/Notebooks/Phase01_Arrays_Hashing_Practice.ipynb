## 1. Two Sum (LeetCode #1)
| Attribute | Specification |
| :--- | :--- |
| **Difficulty** | 🟢 Easy | **Priority** | ⭐ Core |
| **Patterns** | `#hash-map` `#complement-lookup` `#one-pass` |

---

### ⚡ 30-Second Executive Pattern Flashcard
- **Trigger Signal:** Unsorted array + find pair summing to target value.
- **Governing Invariant:** For current cursor $i$, its complement $(target - nums[i])$ must already exist in historical ledger $seen$.
- **Subtle Trap:** Self-matching duplicate values. Must probe $seen$ *before* inserting $nums[i]$ to prevent using the same index twice.
- **Complexity:** **Time:** $O(N)$ amortized | **Auxiliary Space:** $O(N)$ hash table | **Output Space:** $O(1)$

---

### 🎙️ Senior Interviewer Dialogue Script ("What to Say Out Loud")
1. **The Lead-In (Clarify & Baseline):**
   > *"We are tasked with finding two distinct indices whose values sum to target. The naive baseline compares every pair $(i, j)$ using two nested loops, taking $O(N^2)$ time and $O(1)$ space. The computational bottleneck is redundant forward scanning with zero historical memory."*
2. **The Invariant Pivot:**
   > *"By inverting $a + b = target \iff b = target - a$, we convert an uncertain $O(N)$ future search into an $O(1)$ historical lookup. We maintain a hash map mapping each seen number to its index."*
3. **Senior Defense (Scale & Systems):**
   > *"In C#, we pre-allocate the Dictionary with capacity $N$ to prevent dynamic rehashing and GC pressure. For distributed streaming data (e.g. transactions in Azure/AWS), we would partition events by key modulus or use a bounded sliding window cache with TTL."*



````python
# 💻 PRACTICE CELL: Problem #1 - Two Sum
# INVARIANT: seen contains only indices < i. Probe complement BEFORE adding nums[i].

def two_sum(nums: list[int], target: int) -> list[int]:
    # TODO: Implement your solution here
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

````

````python
# 🧪 TEST ASSERTION CELL: Run to verify your solution
assert two_sum([2, 7, 11, 15], 9) == [0, 1], "Test 1 Failed"
assert two_sum([3, 2, 4], 6) == [1, 2], "Test 2 Failed"
assert two_sum([3, 3], 6) == [0, 1], "Test 3 (Duplicates) Failed"
assert two_sum([-1, -2, -3, -4, -5], -8) == [2, 4], "Test 4 (Negatives) Failed"
assert two_sum([0, 4, 3, 0], 0) == [0, 3], "Test 5 (Zeros) Failed"
print("✅ All Two Sum tests passed!")

````

```csharp
    public static int[] TwoSum(int[] nums, int target) {
        ArgumentNullException.ThrowIfNull(nums);
        var seen = new Dictionary<int, int>(capacity: nums.Length);

        for (int i = 0; i < nums.Length; i++) {

            if (seen.TryGetValue(complement, out int prevIdx)) {
                return [prevIdx, i];
            }
            seen[nums[i]] = i;
        }
        return [];
    }
}
```

## 4. Group Anagrams (LeetCode #49)
| Attribute | Specification |
| :--- | :--- |
| **Difficulty** | 🟡 Medium | **Priority** | ⭐ Core |
| **Patterns** | `#hash-map` `#canonical-representation` `#frequency-array` |

---

### ⚡ 30-Second Executive Pattern Flashcard
- **Trigger Signal:** Group collections of items by equivalence under reordering / permutation.
- **Governing Invariant:** All anagrams share an identical **canonical signature**. Map `signature -> list[string]`.
- **Subtle Trap:** Sorting each string costs $O(K \log K)$ per word ($O(N \cdot K \log K)$ total). For high throughput, use a 26-element character count tuple as the canonical key to achieve strictly linear $O(N \cdot K)$ runtime.
- **Complexity:** **Time:** $O(N \cdot K)$ | **Auxiliary Space:** $O(N \cdot K)$ hash table | **Output Space:** $O(N \cdot K)$

---

### 🎙️ Senior Interviewer Dialogue Script ("What to Say Out Loud")
1. **The Lead-In (Clarify & Baseline):**
   > *"We need to group strings that contain identical character frequencies. Sorting each word yields an $O(N \cdot K \log K)$ runtime. The bottleneck is redundant comparison between words that belong to different equivalence classes."*
2. **The Invariant Pivot:**
   > *"Instead of pairwise sorting, we define an immutable canonical representation: a 26-element character frequency array. Since two strings are anagrams if and only if their frequency counts are identical, this tuple acts as an $O(K)$ hash key."*
3. **Senior Defense (Scale & Systems):**
   > *"This is the exact pattern used in distributed MapReduce word bucketing and document similarity clustering. In distributed systems (e.g. Azure CosmosDB / AWS DynamoDB), we partition documents across worker nodes using consistent hashing on this canonical signature."*



````python
# 💻 PRACTICE CELL: Problem #4 - Group Anagrams
# INVARIANT: Map canonical 26-int tuple signature to list of words.

def group_anagrams(strs: list[str]) -> list[list[str]]:
    # TODO: Implement your solution here
    from collections import defaultdict
    groups = defaultdict(list)
    for s in strs:
        count = [0] * 26
        for ch in s:
            count[ord(ch) - ord('a')] += 1
        groups[tuple(count)].append(s)
    return list(groups.values())

````

````python
# 🧪 TEST ASSERTION CELL: Run to verify your solution
res1 = group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
res1_sorted = sorted([sorted(g) for g in res1])
expected1 = sorted([sorted(g) for g in [["bat"], ["nat", "tan"], ["ate", "eat", "tea"]]])
assert res1_sorted == expected1, f"Failed: {res1_sorted} != {expected1}"

assert group_anagrams([""]) == [[""]], "Failed empty string"
assert group_anagrams(["a"]) == [["a"]], "Failed single string"
print("✅ All Group Anagrams tests passed!")

````

```csharp
public static class Solution {
    public static IList<IList<string>> GroupAnagrams(string[] strs) {
        if (strs == null || strs.Length == 0) return [];
        var map = new Dictionary<string, List<string>>(strs.Length);

        Span<char> count = stackalloc char[26];
        foreach (var s in strs) {
            count.Clear();
            foreach (char c in s) count[c - 'a']++;
            string key = new string(count); // string allocation safe for dictionary key
            if (!map.TryGetValue(key, out var list)) {
                list = new List<string>();
                map[key] = list;
            }
            list.Add(s);
        }
        return map.Values.Cast<IList<string>>().ToList();
    }
}
```
