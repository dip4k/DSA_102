## 10. Container With Most Water (LeetCode #11)
| Attribute | Specification |
| :--- | :--- |
| **Difficulty** | 🟡 Medium | **Priority** | ⭐ Core |
| **Patterns** | `#two-pointers` `#greedy-elimination` `#boundary-shrink` |

---

### ⚡ 30-Second Executive Pattern Flashcard
- **Trigger Signal:** Maximize rectangular area between two walls: $\text{Area} = (R - L) \times \min(H[L], H[R])$.
- **Governing Invariant:** The shorter of the two walls strictly limits capacity. Moving the taller wall inward can *only decrease or maintain width while never increasing the height limiter*. Therefore, the shorter wall can be safely discarded forever.
- **Subtle Trap:** Attempting to move both pointers or moving the taller pointer looking for a higher wall (violates discard safety).
- **Complexity:** **Time:** $O(N)$ | **Auxiliary Space:** $O(1)$ | **Output Space:** $O(1)$

---

### 🎙️ Senior Interviewer Dialogue Script ("What to Say Out Loud")
1. **The Lead-In (Clarify & Baseline):**
   > *"We want to find two lines that together with the x-axis form a container holding the maximum water. The naive baseline tests all $N(N-1)/2$ pairs in $O(N^2)$ time. The bottleneck is evaluating redundant internal pairs bounded by a known shorter wall."*
2. **The Invariant Pivot:**
   > *"Width strictly decreases as we move inward. The area is constrained by $\min(height[L], height[R])$. If we keep the shorter wall and move the taller wall, width decreases and height cannot exceed the current shorter wall—meaning area is guaranteed to shrink. Hence, moving the shorter wall is the only move that can possibly discover a larger area."*
3. **Senior Defense (Scale & Systems):**
   > *"This is a classic greedy boundary elimination in $O(1)$ auxiliary memory and cache-friendly contiguous array scanning. In stream processing (e.g. dynamic SLA envelopes in cloud telemetry), this boundary contraction enables real-time peak-capacity calculation in a single pass."*



````python
# 💻 PRACTICE CELL: Problem #10 - Container With Most Water
# INVARIANT: Advance the pointer pointing to the shorter wall.

def max_area(height: list[int]) -> int:
    # TODO: Implement your solution here
    left, right = 0, len(height) - 1
    max_water = 0
    while left < right:
        h_left, h_right = height[left], height[right]
        width = right - left
        current_area = min(h_left, h_right) * width
        if current_area > max_water:
            max_water = current_area
        if h_left < h_right:
            left += 1
        else:
            right -= 1
    return max_water

````

public static class Solution {
    public static int MaxArea(int[] height) {
        ArgumentNullException.ThrowIfNull(height);
        int left = 0, right = height.Length - 1;
        int maxWater = 0;

        while (left < right) {
            int width = right - left;
            int hL = height[left];
            int hR = height[right];

            int currentArea = (hL < hR ? hL : hR) * width;
            if (currentArea > maxWater) maxWater = currentArea;

            if (hL < hR) left++;
            else right--;
        }
        return maxWater;
    }
}

## 13. Trapping Rain Water (LeetCode #42)
| Attribute | Specification |
| :--- | :--- |
| **Difficulty** | 🔴 Hard | **Priority** | ⭐ Core / Flagship Anchor |
| **Patterns** | `#two-pointers` `#prefix-suffix-max` `#boundary-invariant` |

---

### ⚡ 30-Second Executive Pattern Flashcard
- **Trigger Signal:** Non-negative elevation map, calculate trapped surface water over irregular terrain.
- **Governing Invariant:** Water at index $i$ depends exclusively on $\min(\text{leftMax}[i], \text{rightMax}[i]) - height[i]$. If $\text{leftMax} < \text{rightMax}$, the water level at $left$ is unconditionally dictated by $\text{leftMax}$, regardless of any internal heights.
- **Subtle Trap:** Allocating two $O(N)$ prefix/suffix arrays. The two-pointer approach compresses auxiliary space from $O(N)$ down to $O(1)$.
- **Complexity:** **Time:** $O(N)$ | **Auxiliary Space:** $O(1)$ | **Output Space:** $O(1)$

---

### 🎙️ Senior Interviewer Dialogue Script ("What to Say Out Loud")
1. **The Lead-In (Clarify & Baseline):**
   > *"The amount of water above bar $i$ is determined by the minimum of the highest wall to its left and highest wall to its right, minus $height[i]$. A naive brute-force scans left and right for every bar, taking $O(N^2)$ time. A dynamic programming approach precomputes prefix and suffix max arrays in $O(N)$ time and $O(N)$ space."*
2. **The Invariant Pivot (Space Optimization to $O(1)$):**
   > *"We can eliminate the $O(N)$ space entirely using two converging pointers. If `leftMax < rightMax`, we know with 100% mathematical certainty that the water above `left` is bounded by `leftMax`, because whatever unvisited terrain exists between `left` and `right`, it cannot drop below `rightMax` (which is already larger than `leftMax`). Hence, we compute water at `left` and advance `left++` safely."*
3. **Senior Defense (Scale & Systems):**
   > *"This achieves zero allocations on the heap, single-pass $O(N)$ execution, and optimal CPU pipeline cache locality. In financial systems, this exact boundary invariant is used for calculating cumulative drawdown buffers between peak orderbook levels."*



````python
# 💻 PRACTICE CELL: Problem #13 - Trapping Rain Water
# INVARIANT: If left_max < right_max, water at left is left_max - height[left].

def trap(height: list[int]) -> int:
    # TODO: Implement your solution here
    if not height or len(height) < 3:
        return 0
    left, right = 0, len(height) - 1
    left_max, right_max = height[left], height[right]
    trapped_water = 0

    while left < right:
        if left_max < right_max:
            left += 1
            if height[left] < left_max:
                trapped_water += left_max - height[left]
            else:
                left_max = height[left]
        else:
            right -= 1
            if height[right] < right_max:
                trapped_water += right_max - height[right]
            else:
                right_max = height[right]
    return trapped_water

````

public static class Solution {

        if (height == null || height.Length < 3) return 0;

        int left = 0, right = height.Length - 1;

                    leftMax = height[left];

                if (height[right] < rightMax) {
                    totalWater += rightMax - height[right];
                } else {
                    rightMax = height[right];
                }
            }
        }
        return totalWater;
    }
}
